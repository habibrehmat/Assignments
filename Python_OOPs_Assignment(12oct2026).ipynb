**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2
class BankAccount():
    def __init__(self, title, balance):
        self.title=title
        self.__balance=balance

    def get_balance(self):
        return print(f"balance = {self.__balance}")
        
    def deposit(self,amount):
        self.__balance+=amount

    def withdraw(self, amount):
        if self.__balance>amount:
            self.__balance-=amount
            return print(f"you have withdraw {amount}  successfully")
        else:
            return print("Insufficient Balance ")



b1=BankAccount("Ali",1000)
b1.get_balance()
b1.deposit(500)
b1.get_balance()
b1.withdraw(500)
b1.get_balance()
b1.withdraw(2500)
b1.get_balance()
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Write your code here for Q10
class shape():
    def __init__(self, colour):
        self.colour=colour
    
class circle(shape):
    def __init__(self, colour, radius, diameter):
        super().__init__(colour)
        self.radius=radius
        self.diameter=diameter
    def area(self,radius):
        area= radius*radius*22/7
        return area
    
class square(shape):
    def __init__(self, colour, side):
        super().__init__(colour)
        self.side=side

    def area(self,side):
        area=side*side
        return area


s1=square("red", 10)
print(f"area of square  is {s1.area(s1.side)}")
c1=circle("yellow", 3,6)
print(f"area of circle  is {c1.area(c1.radius)}")



    
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13
#operator overloading: same method name with different arguments withen the class, and both are working seperately. python do not support it, python uses only the latest one method.
class point():
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __add__(self,self2):
        x2=(self.x + self2.x)
        y2=(self.y + self2.y)
        return point (x2,y2)
    def __str__(self):
        return f" {self.x}, {self.y}"

p1=point(2,3)
p2=point(4,7)
p3= p1 + p2
print("sum point",p3)
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
from abc import ABC, abstractmethod

#abstract parent Class
class Payment(ABC):
    def __init__(self):
        pass

    #abstract method, every child class has to write its own version
    @abstractmethod
    def pay(self,amount):
        self.amount=amount
        pass

#first Child class:CreditCard
class CreditCard(Payment):
    def pay(self,amount):
        self.amount=amount
        print (f"paid {self.amount} by using CreditCard")

#second Child class: UPI
class UPI(Payment):
        def pay(self,amount):
            self.amount=amount
            print (f"paid {self.amount} by using UPI")


#test the working

p1=CreditCard()
p2=UPI()

p1.pay(1300)
p2.pay(400)
    
```
