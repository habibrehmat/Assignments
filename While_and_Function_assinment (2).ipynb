## Part 1: While Loop


**Q1.** Ek while loop banayein jo 'Hello' ko 3 martaba print kare.

```python
# Apna code yahan likhein
i = 0   #initializing integer i
while i<3:      #run following block while i is less than 3
    print("Hello")# print word "Hello"
    i+=1       #add increment in i on each iteration
```

**Q2.** 10 se 1 tak ulti ginti (reverse counting) print karne ka while loop likhein.

```python
# Apna code yahan likhein
i = 10   #initializing integer i = 10 i.e. initial value
while i>0:      #run following block while i is greater than 0
    print(i)# print number i
    i-=1       #add increment (-1) in i on each iteration
```

**Q5.** Ek infinite loop banayein aur `break` statement use kar ke usay foran rok dein.

```python
# Apna code yahan likhein
n=5
i=1
while i<11:    # run loop untill i<11
    print(f" {n} x {i} = {n*i}")
    i-=1
    if i==-5: #check i== -5
        print("while loop intrupted by 'break'")
        break  # break the loop when above condition is true

```

**Q4.** Ek function `find_maximum(a, b, c)` banayein jo 3 numbers mein se sabse bada number bataye.

```python
# Apna code yahan likhein
def find_maximum(a,b,c):
    max=a
    for i in [b,c]:
        if max<i:
            max=i
    return max
find_maximum(2,4,11)
```

**Q9.** Ek function `calculate_factorial(n)` banayein jo number ka factorial return kare.

```python
# Apna code yahan likhein
def calculate_factorial(n):
    out=1
    while n>=1:
        out=out*(n)
        n-=1
    return out
calculate_factorial(5)

```

**Q10.** Ek function `count_vowels(text)` banayein jo string mein vowels (a, e, i, o, u) ki tadad bataye.

```python
# Apna code yahan likhein
def count_vowels(text):
    count=0
    for i in text:
        if i in ["a","e","i","o","u"]:
            count+=1
    return count

count_vowels("habib")
```
