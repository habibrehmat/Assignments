### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
# Q19 solution code here
def valid_age(age):
    if age<18:
        raise ValueError ("Age must be 18 or older")
    print("Age is ok")
try:
    age=int(input("apni age Enter karain"))
    valid_age(age)
except ValueError as error:
    print(f"hamara errpr: {error}")
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
# Q20 solution code here
#
try:
    file=open("number.txt","r")
    print(file.read())
except FileNotFoundError:
    print("file not exist")
except exception as e:
    print(f"file exist, but error{e}")
finally:
    if file:
        file.close()
        print("file closed using 'Finaly' Block")
    else:
        print("no need to close file because file not exist")

```
